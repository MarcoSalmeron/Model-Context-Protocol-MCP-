# Patron de razonamiento: ReAct (Razonar, Actuar, Observar)
### Razonar paso a paso y con el uso de tools 

```python
from dotenv import load_dotenv
import os
load_dotenv(override=True)
print(bool(os.getenv("OPENAI_API_KEY")))
```

### LLM

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("openai:gpt-4o-mini", temperature=0.0)
```

### Tools

```python
from langchain_core.tools import tool

@tool
def multiply(x: int, y: int) -> int:
    """Multiply x by y"""
    return x * y

@tool
def divide(x: int, y: int) -> float:
    """Divide x by y"""
    return x / y

@tool
def add(x: int, y: int) -> int:
    """Add x and y"""
    return x + y
```

### LangChain Agent

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[multiply, divide, add],
)
```

```python
def run_agent(agent, prompt):
    result = agent.invoke({"messages": [{"role": "user", "content": prompt}]})
    
    for msj in result['messages']:
        print('#'*30)
        print(f"[{msj.type.upper()}]: {msj.content}")
        print('#'*30)
```

```python
run_agent(agent, "cuannto es la multiplicacion de 2 por 5 menos 3 y la suma de 3 y 4")
```
